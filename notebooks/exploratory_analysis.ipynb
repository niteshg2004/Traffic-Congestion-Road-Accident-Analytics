# Exploratory analysis of Spark results
Run the pipeline first (`python -m src.run_pipeline`). This notebook only reads the small aggregated tables in `results/tables/` with Pandas (no Spark).
Set `SAMPLE = True` only to test with the synthetic data.

```python
import json, pandas as pd, plotly.express as px
from pathlib import Path
SAMPLE = False
T = Path('..')/('results_sample' if SAMPLE else 'results')/'tables'
print(T, T.exists())
```

```python
print(json.loads((T/'kpis.json').read_text()))
```

```python
by_hour = pd.read_csv(T/'by_hour.csv')
px.bar(by_hour, x='Hour', y='accidents', title='Accidents by hour')
```

```python
by_state = pd.read_csv(T/'by_state.csv')
px.bar(by_state.head(15), x='State', y='accidents', title='Top 15 states')
```
